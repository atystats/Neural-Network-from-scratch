# Coding a neuron

## A single neuron

Imagine 3 input variable with a single training example. 
We are building a single neuron.

```python
input = [1, 2, 3]
weights = [0.2, 0.7, -0.5]
bias = 2
```

```python
output = (input[0] * weights[0]) + (input[1] * weights[1]) + (input[2] * weights[2]) +  bias
```

```python
print(output)
```

## A layer of neuron
Let's assume that we have a layer of neuron with few neurons.

```python
input = [1, 2, 2.5, 3]

# weights for each neuron in the layer
weights1 = [0.2, 0.8, -0.45, 0.7]
weights2 = [0.5, -0.9, 0.26, -0.5]
weights3 = [-0.26, -0.27, 0.17, 0.87]

#bias for each neuron in the layer

bias1 = 2
bias2 = 4
bias3 = 0.8
```

```python
#output will be a list of numbers with length equals to no of neurons
output = [(input[0] * weights1[0]) + (input[1] * weights1[1]) + (input[2] * weights1[2]) + (input[3] * weights1[3]) + bias1,
          (input[0] * weights2[0]) + (input[1] * weights2[1]) + (input[2] * weights2[2]) + (input[3] * weights2[3]) + bias2,
          (input[0] * weights3[0]) + (input[1] * weights3[1]) + (input[2] * weights3[2]) + (input[3] * weights3[3]) + bias3,]

print(output)
```

Each neuron has its own weights, and each neuron is connected to same inputs. This is called a fully connected neural network.

As the number of neurons increase, it can very lengthy code so we are going to use a for loop.

```python
weights = [[0.2, 0.8, -0.45, 0.7],
           [0.5, -0.9, 0.26, -0.5],
           [-0.26, -0.27, 0.17, 0.87]]

bias = [2, 4, 0.8]
```

```python
layer_output = []
for neuron_weights, neuron_bias in zip(weights, bias):
    neuron_output = 0
    for weight, inp in zip(neuron_weights, input):
        neuron_output += weight * inp

    neuron_output += neuron_bias
    layer_output.append(neuron_output)
```

```python
print(layer_output)
```

If we consider weights and inputs as vectors, then the output of neuron to an input is called a dot product of weights and input. These dot products can be calculated using numpy.

## A single neuron with numpy

```python
import numpy as np

input = [1, 2, 3]
weights = [0.2, 0.7, -0.5]
bias = 2
```

```python
output = np.dot(input, weights) + bias
```

```python
print(output)
```

## A layer of neuron using numpy

```python
input = [1, 2, 2.5, 3]

weights = [[0.2, 0.8, -0.45, 0.7],
           [0.5, -0.9, 0.26, -0.5],
           [-0.26, -0.27, 0.17, 0.87]]

bias = [2, 4, 0.8]
```

```python
output = np.dot(weights, input) + bias
```

```python
print(output)
```

Neural network get data in batches, so now we will build a layer of neuron with batch of data with Numpy.

## A layer of neuron & a batch of data with Numpy

```python
input = [[1, 2, 3, 2.5],
         [2, 5, -1, 2],
         [-1.5, 2.7, 3.3, -0.8]]

weights = [[0.2, 0.8, -0.45, 0.7],
           [0.5, -0.9, 0.26, -0.5],
           [-0.26, -0.27, 0.17, 0.87]]

bias = [2, 4, 0.8]
```

```python
output = np.dot(input, np.array(weights).T) + bias
```
