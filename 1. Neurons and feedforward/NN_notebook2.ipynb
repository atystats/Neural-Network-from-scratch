# Deep Neural Network

A neural network with 2 or more hidden layer is called a deep neural network.

We are assuming 3 units in 2 layers of NN. 

Our input batch has 4 features and 3 sample. Hence our input matrix is 3X4. Our 1st layer of weights would be 3X4, which we will transpose for matrix multiplication. Our 2nd layer will also be 3X4. 

Input X weights1^T * weights2^t

```python
import numpy as np
from nnfs.datasets import spiral_data
import nnfs
import matplotlib.pyplot as plt
```

```python
input = [[1, 2, 3, 2.5],
         [2, 5, -1, 2],
         [-1.5, 2.7, 3.3, -0.8]]

weights1 = [[0.2, 0.8, -0.45, 0.7],
           [0.5, -0.9, 0.26, -0.5],
           [-0.26, -0.27, 0.17, 0.87]]

bias1 = [2, 4, 0.8]
```

```python
weights2 = [[0.1, -0.14, 0.5],
            [-0.5, 0.12, -0.33],
            [-0.44, 0.73, -0.13]]

bias2 = [-1, 2, -0.5]
```

```python
layer1_output = np.dot(input, np.array(weights1).T) + bias1
```

```python
print(layer1_output)
```

```python
layer2_output = np.dot(layer1_output, np.array(weights2).T) + bias2
```

```python
print(layer2_output)
```

## Using a data generator to create input.

```python
X, y = spiral_data(samples = 100, classes = 3)
plt.scatter(X[:,0], X[:,1], c= y, cmap = 'brg')
plt.show()
```

Creating a class for Dense Layer of Neural Network

```python

class Dense_layer:

    def __init__(self, n_inputs, n_neurons):
        self.weights = 0.01 * np.random.randn(n_inputs, n_neurons)
        self.biases = np.zeros((1, n_neurons))
    
    def forward(self, inputs):
        self.output = np.dot(inputs, self.weights) + self.biases
```

```python
dense1 = Dense_layer(2,3)

dense1.forward(X)

print(dense1.output[:5])
```
